## 2. The question: decision, action, cost of a wrong call

**Research Question**: How can we predict and rank content items undergoing search traffic decline to prioritize them for updates?

* **Unit of Analysis**: A single content item (`content_hash_id`) over a trailing 90-day feature window.
* **Decision Improved**: "Which page should an editor review and update first?" (translating thousands of pages into a prioritized queue).
* **Who Acts & Action**: An EdTech content editor or SEO manager. Their action is to perform a content audit on the ranked page, update outdated details, add missing depth, and re-publish.
* **Cost of a Wrong Call**:
  * *False Positive (Wasted Effort)*: Flagging a stable page that cannot be improved by editing. This wastes valuable writing hours.
  * *False Negative (Missed Decline)*: Failing to identify a key traffic-driver in decline, causing it to fall off SERP Page 1.
* **Why ML helps**: Simple rules like "flag pages older than 6 months" are highly inefficient. ML can weigh multiple variables—like historical impressions, position drift, CTR decay, and user engagement—to identify decay before traffic collapses.

## 3. Quick look at the data (2-3 real numbers)

Analyzing the `data/raw/content_refresh_anonymized.csv` starter dataset yields the following supporting numbers:

1. **Total Decline Rate**: **54.21%** of pages in the dataset are declining.
2. **Impressions at Risk**: **51.27%** of all search visibility (impressions) is concentrated on pages currently in decline (79.9M out of 156M impressions).
3. **Stale vs. Fresh Page Decline Rate**: The decline rate of "fresh" pages (updated < 180 days ago) is **54.25%**, which is actually *higher* than "stale" pages (updated >= 180 days ago) at **47.13%**. This proves a simple time-since-update rule is highly inefficient and would miss more than half of the declining pages.

```python
import pandas as pd
import numpy as np

# Load the starter CSV
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining"] = df["trend_direction"].str.lower().eq("down")

total_rows = df.shape[0]
decline_rate = df["is_declining"].mean()

stale_mask = df["days_since_last_update"] >= 180
stale_decline_rate = df[stale_mask]["is_declining"].mean()
fresh_decline_rate = df[~stale_mask]["is_declining"].mean()

total_impressions = df["impressions_90d"].sum()
declining_impressions = df[df["is_declining"]]["impressions_90d"].sum()
impression_risk_pct = declining_impressions / total_impressions

print(f"Total Rows: {total_rows:,}")
print(f"Decline Rate: {decline_rate*100:.2f}%")
print(f"Stale Pages Decline Rate: {stale_decline_rate*100:.2f}%")
print(f"Fresh Pages Decline Rate: {fresh_decline_rate*100:.2f}%")
print(f"Impressions on Declining Pages: {declining_impressions:,} / {total_impressions:,} ({impression_risk_pct*100:.2f}%)")
```
